```python
import numpy as np
import random
```

```python
na=200
nb=50

P = np.random.rand(na, nb)
print(P)
```

```python
x=2
y=2
grado_a= np.sum(P, axis=1)
grado_b=np.sum(P, axis=0)

P1=np.copy(P)
P1[x,y]=1
for j in range(nb):
    if j==y:
        continue
    P1[x,j]=P[x,j]*((grado_a[x]-1)/(grado_a[x]-P[x,y]))

for i in range(na):
    if i==x:
        continue
    P1[i,y]=P[i,y]*((grado_b[y]-1)/(grado_b[y]-P[x,y]))
P2=np.zeros((na,nb))
P2[x,:]=P1[x,:]
P2[:,y]=P1[:,y]
grado_a1=np.sum(P1, axis=1)
grado_b1=np.sum(P1, axis=0)
delta_k_a=grado_a-grado_a1
delta_k_b=grado_b-grado_b1
gamma=np.sum(delta_k_a)+np.sum(delta_k_b)
sigma=0
for i in range(na):
    temp=delta_k_a[i]*(grado_a[i]-P[i,y])
    sigma=sigma+temp
for i in range(nb):
    temp=delta_k_b[i]*(grado_b[i]-P[x,i])
    sigma=sigma+temp
for i in range(na):
    for j in range(nb):
        if i==x or j==y:
            continue

        P2[i,j]=P[i,j]*(1+(gamma/sigma)*(delta_k_a[i]+delta_k_b[j]))
grado_a2=np.sum(P2, axis=1)
grado_b2=np.sum(P2, axis=0)
print(grado_b2-grado_b)
```
